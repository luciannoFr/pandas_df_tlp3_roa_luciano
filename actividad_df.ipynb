```python
import pandas as pd
import sqlite3

#Aqui agrega la conexión a la base de datos

conn = sqlite3.connect('Salaries.sqlite')

#Crea el dataframe con los datos del datasets

df = pd.read_sql_query("SELECT * FROM salaries", conn)
df.head()
```

1. **Top 10 empleados con mayor salario total (incluyendo beneficios)**
Extrae e imprime un top 10 de expleados con mayor salario incluyendo los beneficios 


```python
#todo

# Obtener el top 10 empleados con mayor salario total
top_10_empleados = df.sort_values(by='TotalPayBenefits', ascending=False).head(10)

# mostramos el top 10
top_10_empleados
```

2. **Filtrar empleados con más de 50,000 en horas extra, la columna a tratar es: (OvertimePay)**

```python
#todo

# cambia textos no numéricos por NaN
df['OvertimePay'] = pd.to_numeric(df['OvertimePay'], errors='coerce')

# filtrar los empleados con mas de 50k en pagos extra
df_filtrado = df[df['OvertimePay'] > 50000].sort_values(by='OvertimePay', ascending=False)

# muestro los primeros 10
df_filtrado.head(10)
```

3. **Contar cuántos empleados únicos hay por año**

```python
#todo
# agrupa por año y luego contamos el registro unico en empleados de cada año
df.groupby('Year')['EmployeeName'].nunique()
```

4. **Ver cuántos cargos únicos (JobTitle) existen y los 5 más comunes**

```python
#todo

# contar cuantos cargos unicos hay
num_cargos_unicos = df['JobTitle'].nunique()

#trae los mas comunes y pone los primeros 5
top5_cargos = df['JobTitle'].value_counts().head(5)

print("cargos unicos:", num_cargos_unicos)
print("top 5 cargos mas comunes:\n", top5_cargos)
```

5. **Mostrar el salario total (TotalPay) promedio por año**

```python
#todo

# calculamos el salario total promedio por cada año
salario_promedio_por_año = df.groupby('Year')['TotalPay'].mean().round(2)

# mostramos los resultados
print(salario_promedio_por_año)
```
